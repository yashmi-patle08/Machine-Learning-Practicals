```python
features=['fixed acidity', 'volatile acidity', 'citric acid', 'residual sugar',
       'chlorides', 'free sulfur dioxide', 'total sulfur dioxide', 'density',
       'pH', 'sulphates', 'alcohol']
for col in features:
  plt.figure(figsize=(7,4))

  sns.boxplot(x='quality',y=col,data=df)
  plt.title(f'{col} vs Wine Quality')
  plt.show()
```

```python
features=df.drop(columns=['quality'])
Q1=features.quantile(0.25)
Q3=features.quantile(0.75)
IQR=Q3-Q1
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
outlier_mask=(features<lower_bound)|(features>upper_bound)
outlier_count=outlier_mask.sum()
outlier_percentage=(outlier_count/len(features))*100
outlier_summary=pd.DataFrame({
    'Q1':Q1,
    'Q3':Q3,
    'IQR':IQR,
    'Lower Bound':lower_bound,
    'Upper Bound':upper_bound,
    'Outlier Count':outlier_count,
    'Outlier %':outlier_percentage
})
outlier_summary.round(3)
```

```python
plt.figure(figsize=(12,6))
outlier_count.sort_values(ascending=False).plot(kind='bar')
plt.title("Number of Outlier in Each Features")
plt.xlabel("Features")
plt.ylabel("Number of Outliers")
plt.xticks(rotation=45,ha='right')
plt.grid(axis='y',alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
y=df['quality']
```

```python
x=df.drop(columns=['quality','quality_group'],errors='ignore')
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42,stratify=y)
```

```python
print("X train:",x_train.shape)
print("X test:",x_test.shape)
print("Y train:",y_train.shape)
print("Y test:",x_test.shape)
```

```python
dt_model=DecisionTreeClassifier(criterion=criterion,random_state=42)
dt_model.fit(x_train,y_train)
```

```python
dt_model.fit(x_train,y_train)
```

```python
y_train_pred=dt_model.predict(x_train)
y_test_pred=dt_model.predict(x_test)
```

```python
train_accuracy=accuracy_score(y_train,y_train_pred)
print("Training Accuracy:",train_accuracy)
```

```python
test_accuracy=accuracy_score(y_test,y_test_pred)
print("Testing Accuracy:",test_accuracy)
```

```python
print("Precision:",precision_score(y_test,y_test_pred,average='weighted'))
print("Recall:",recall_score(y_test,y_test_pred,average='weighted'))
print("f1_score:",f1_score(y_test,y_test_pred,average='weighted'))
```

```python
print(classification_report(y_test,y_test_pred))
```

```python
cm=confusion_matrix(y_test,y_test_pred)
plt.figure(figsize=(8,6))
sns.heatmap(cm,annot=True,fmt='d',cmap='Blues')
plt.xlabel("Predicted")
plt.ylabel("Actual Quality")
plt.title("Confusion Matrix - Decision Tree")
plt.show()
```

```python
print("Tree depth:",dt_model.get_depth())
print("Number of leaves:",dt_model.get_n_leaves())
```

```python
plt.figure(figsize=(25,15))
plot_tree(
    dt_model,
    feature_names=x.columns,
    class_names=[str(c) for c in sorted(y.unique())],
    filled=True,
    rounded=True,
    fontsize=8
)
plt.title("Decision Tree - Entropy")
plt.show()
```
